```python
# SIMPLE CALCULATOR:

num1 = int(input("please enter num1"))
num2 = int(input("please enter num2"))

print("options:\n", "+","-","*","/")

choice = input("please enter any one choice")

if choice == "+":
    print("sum is:\n", num1+num2)
elif choice == "-":
    print("substraction is:\n", num2- num1)
elif choice == "*":
    print("product is:\n", num1*num2)
elif choice == "/":
    print("division is:\n", num1/num2)
else:
    print("somethink wrong")
    
        
```

```python
# salary increment:

salary = float(input("Enter your current salary:"))
bonus_percentage = float(input("Enter your bonus percentage:"))

bonus_amount = salary * (bonus_percentage / 100)

salary += bonus_amount

print("Bonus amount:", bonus_amount)
print("Updated salary:", salary)


```

```python
# gym membership fee calculator:

registration_fee = int(input("enterregistration fees"))
monthly_fees = int(input("enter monthly fees"))
no_of_month = int(input("enter number of month you want pay"))

total_fees = registration_fee+(monthly_fees*no_of_month)
print(total_fees)

discount_coupon = input("have discout coupon, yes or not?").lower()
if discount_coupon == "yes":
    total_fees *= 0.9
    print("final amount", total_fees)
    
    
                        
```
